### SQL Database Discovery — Quickstart

```python
#imports 
import pandas as pd
from sqlalchemy import create_engine
from urllib.parse import urlparse, quote_plus
import requests
import pyodbc
import os
```

We set connection parameters for our database and create connection string. While it is not important which particular database you connect, it you can use “postgres” database for convenience:

```python
# set credentials, server and database name

username = 'my_user'
password = 'my_password'
server = 'my_server'
port = 'my_port'
database = 'my_db'

# create connection string using the credentials

conn_string = 'postgresql://{}:{}@{}:{}/{}'.format(username,password,server, port, database)
db = create_engine(conn_string)
conn = db.connect()
```

Then, we need to provide a query that gets list of databases from our instance. We will use this list later:

```python
# provide query to get list of databeses 

query = 'SELECT datname FROM pg_database WHERE datistemplate = false;'

db_list = pd.read_sql(query, conn)


# create new list containing database names  

dblist = db_list['datname'].to_list()
```

Once we have the list, we can loop through it and get information from each database using below query:

```python
SELECT  current_database() as db_name,
        grantee as user_name, 
        privilege_type as permission, 
        table_schema as schema_name, 
        table_name 
FROM information_schema.role_table_grants
```
